```python
class jugador:
    def __init__(self, poder, nombre, vidas, poder_especial, nivel, armamento):
        self.poder = poder
        self.nombre = nombre
        self.vidas = vidas
        self.poder_especial = poder_especial
        self.nivel = nivel
        self.armamento = armamento

ninja1  = jugador( 333, "nicolas maduro", 3, "vengan por mi cobardes", "medio", "bajo" )
ninja2  = jugador( 798, "chaves", 3, "habilidades del rey esqueleto y aguantar dias sin comer", "medio", "alto")


class enemigo:
    def __init__(self, poder, nombre, vidas, poder_especial, nivel, armamento):
        self.poder = poder
        self.nombre = nombre
        self.vidas = vidas
        self.poder_especial = poder_especial
        self.nivel = nivel
        self.armamento = armamento

dragon = enemigo( 9999, "donald trump", 7, "Bombardeo Armagedón", "extremo", "extremo")

print("la informacion del primer ninja es: ")
print(f"él poder es: {ninja1.poder}")
print(f"él nombre es: {ninja1.nombre}")
print(f"sus vidas son: {ninja1.vidas}")
print(f"él poder especial es: {ninja1.poder_especial}")
print(f"él tamaño es: {ninja1.nivel}")
print(f"él poder es: {ninja1.armamento}")
print("\n")
print("la informacion del segundo ninja es: ")
print(f"él poder es: {ninja2.poder}")
print(f"él nombre es: {ninja2.nombre}")
print(f"sus vidas son: {ninja2.vidas}")
print(f"él poder especial es: {ninja2.poder_especial}")
print(f"él tamaño es: {ninja2.nivel}")
print(f"él poder es: {ninja2.armamento}")
print("\n")
print("la informacion del dragon es: ")
print(f"él poder es: {dragon.poder}")
print(f"él nombre es: {dragon.nombre}")
print(f"sus vidas son: {dragon.vidas}")
print(f"él poder especial es: {dragon.poder_especial}")
print(f"él tamaño es: {dragon.nivel}")
print(f"él poder es: {dragon.armamento}")
```
