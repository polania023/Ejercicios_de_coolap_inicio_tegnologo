```python
from abc import ABC, abstractmethod
#creamos la clase abstract
class GenerarReporte(ABC):
    @abstractmethod
    def generar_reporte(self)-> None:
        pass

class SistemaPago(ABC):
    def __init__(self, monto: float) -> None:
        self.monto = monto

    @abstractmethod
    def procesar_pago(self) -> None:
        pass
class PagoBancario(SistemaPago, GenerarReporte):
    def procesar_pago(self) -> None:
        print(f"Procesando pago bancario de {self.monto:,.2f}")
    def generar_reporte(self) -> None:
        print(f"Reporte generado mediante el sistema bancario")
class PagoCriptomoneda(SistemaPago):
    def procesar_pago(self) -> None:
        print(f"Procesando pago con criptomoneda de {self.monto: ,.2f}")
def ejecutar_pago(pago: SistemaPago) -> None:
        pago.procesar_pago()
class PagoTarjetaCredito(SistemaPago):
    def procesar_pago(self) -> None:
        print(f"Procesando pago con tarjeta de credito de {self.monto: ,.2f}")
    def generar_reporte(self) -> None:
        print(f"Reporte generado mediante el sistema bancario")
def main():
        print("SISTEMA DE PAGOS")
        pago1 = PagoBancario(15.000)
        ejecutar_pago(pago1)
        pago1.generar_reporte()
        print("---------------------------------------")
        pago2 = PagoCriptomoneda(5.0)
        ejecutar_pago(pago2)
        pago1.generar_reporte()
        print("---------------------------------------")
        pago3 = PagoTarjetaCredito(20.000)
        ejecutar_pago(pago3)
        pago3.generar_reporte()

if __name__ == "__main__":
    main()



```
