```python
from abc import ABC, abstractmethod
#creamos la clase abstract
class Empleado(ABC):
  def __init__(self, nombre: str) -> None:
    self._nombre = None # queda como privado y nosa obliga a utulizar getter y setter inicia un vacio
    self.nombre = nombre # nombre ya se puede usar getter y setter
  #CREAMOS EL GETTER
  @property
  def nombre(self) -> str:
    return self._nombre
  #SERTTER
  @nombre.setter
  def nombre(self, valor: str) -> None:#setter profesional con validacion basica
      if isinstance(valor, str) and valor.strip():
        self._nombre = valor.strip()
      else:
        raise ValueError("El nombre ingresado no es u texto valido")
@abstractmethod
def trabajar(self) -> None:
  #Este metodo es el obligatorio para todos los hijos de esta clase
  pass
#todo lo de atras de el padre o la subclase
#ahora crearemos una clase hija
class Gerente(Empleado):
  def trabajar(self) -> None:
      print(f"{self.nombre} esta programado unas funciones en JAVA.\n")

class Desarrollador(Empleado):
  def trabajar(self) -> None:
      print(f"{self.nombre} esta programado unas funciones en JAVA.\n")

def ejecutar_tarea(empleado: Empleado) -> None:# con este metodo un polimorfismo puro
  empleado.trabajar()

def main():
  gerente1 = Gerente("Predrito Perez")
  desarrollador1 = Desarrollador("jorge lluna")
  print("Polimorfismo con Getter")
  ejecutar_tarea(gerente1)
  ejecutar_tarea(desarrollador1)
  #cambiamos el nombre con setter
  gerente1.nombre = "Juanito Rodrigez"
  print(f"El nuevo nombre del gerente es: {gerente1.nombre}")
  ejecutar_tarea(gerente1)
  desarrollador1.nombre = "roone wayne"
  print(f"El nuevo nombre del desarrollador es: {desarrollador1.nombre}")
  ejecutar_tarea(desarrollador1)

if __name__ == "__main__":
  main()
```
