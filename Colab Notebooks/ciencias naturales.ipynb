```python
print("Calculadora MRU y MRUA")

tipo = input("Selecciona el tipo de movimiento (mru / mrua): ").lower()

# Pedir datos
distancia = input("Distancia (m) (dejar vacío si no se sabe): ")
velocidad = input("Velocidad (m/s) (dejar vacío si no se sabe): ")
tiempo = input("Tiempo (s) (dejar vacío si no se sabe): ")
aceleracion = input("Aceleración (m/s²) (dejar vacío si no se sabe): ")

# Convertir a número si existe
d = float(distancia) if distancia != "" else None
v = float(velocidad) if velocidad != "" else None
t = float(tiempo) if tiempo != "" else None
a = float(aceleracion) if aceleracion != "" else None

resultado = ""

if tipo == "mru":
    # d = v * t
    if v is not None and t is not None:
        resultado = f"Distancia = {v * t:.2f} m"

    elif d is not None and t is not None:
        resultado = f"Velocidad = {d / t:.2f} m/s"

    elif d is not None and v is not None:
        resultado = f"Tiempo = {d / v:.2f} s"

    else:
        resultado = "Ingresa al menos dos valores."

elif tipo == "mrua":

    # v = v0 + a*t
    if v is not None and a is not None and t is not None:
        velocidad_final = v + a * t
        resultado = f"Velocidad final = {velocidad_final:.2f} m/s"

        # d = v0*t + 1/2*a*t²
        distancia = v * t + 0.5 * a * t**2
        resultado += f"\nDistancia = {distancia:.2f} m"

    else:
        resultado = "Para MRUA necesitas velocidad inicial, aceleración y tiempo."

else:
    resultado = "Tipo de movimiento no válido."

print("\nResultado:")
print(resultado)
```

```python
from sympy import Matrix, lcm
import re


# ==============================
# 1. Analizar fórmula química
# ==============================
def analizar_formula(formula):
    """
    Convierte una fórmula química en un diccionario con la cantidad de átomos.
    Ejemplo: H2O -> {'H':2, 'O':1}
    """
    patron = r'([A-Z][a-z]?)(\d*)'
    elementos = re.findall(patron, formula)

    composicion = {}

    for elemento, cantidad in elementos:
        cantidad = int(cantidad) if cantidad else 1
        composicion[elemento] = composicion.get(elemento, 0) + cantidad

    return composicion


# ==============================
# 2. Balancear ecuación química
# ==============================
def balancear_ecuacion(reactivos, productos):

    sustancias = reactivos + productos
    composiciones = []
    elementos = set()

    # Analizar cada sustancia
    for sustancia in sustancias:
        comp = analizar_formula(sustancia)
        composiciones.append(comp)
        elementos.update(comp.keys())

    elementos = list(elementos)

    matriz = []

    # Crear matriz de balance
    for elemento in elementos:

        fila = []

        # Reactivos (positivos)
        for comp in composiciones[:len(reactivos)]:
            fila.append(comp.get(elemento, 0))

        # Productos (negativos)
        for comp in composiciones[len(reactivos):]:
            fila.append(-comp.get(elemento, 0))

        matriz.append(fila)

    M = Matrix(matriz)

    try:
        solucion = M.nullspace()[0]
    except:
        print("❌ No se pudo balancear la ecuación")
        return None

    # Convertir fracciones a enteros
    multiplicador = lcm([x.q for x in solucion])
    coeficientes = [abs(int(x * multiplicador)) for x in solucion]

    return coeficientes


# ==============================
# 3. Mostrar ecuación balanceada
# ==============================
def mostrar_ecuacion(reactivos, productos, coeficientes):

    izquierda = []
    derecha = []

    # Reactivos
    for i, r in enumerate(reactivos):
        c = coeficientes[i]
        texto = (str(c) if c != 1 else "") + r
        izquierda.append(texto)

    # Productos
    for i, p in enumerate(productos):
        c = coeficientes[len(reactivos) + i]
        texto = (str(c) if c != 1 else "") + p
        derecha.append(texto)

    return " + ".join(izquierda) + " -> " + " + ".join(derecha)


# ==============================
# 4. Programa principal
# ==============================
def main():

    print("===================================")
    print("   BALANCEADOR DE ECUACIONES")
    print("===================================\n")

    print("Ejemplo de entrada:")
    print("H2 + O2 -> H2O\n")

    ecuacion = input("Ingrese la ecuación química: ").replace(" ", "")

    try:
        izquierda, derecha = ecuacion.split("->")
        reactivos = izquierda.split("+")
        productos = derecha.split("+")
    except:
        print("❌ Error: Formato incorrecto.")
        print("Usa el formato: Reactivos -> Productos")
        return

    coeficientes = balancear_ecuacion(reactivos, productos)

    if coeficientes:
        ecuacion_balanceada = mostrar_ecuacion(reactivos, productos, coeficientes)

        print("\n✅ ECUACIÓN BALANCEADA:")
        print(ecuacion_balanceada)


# Ejecutar programa
main()
```
