```python
class Aprendiz:
    def __init__(self, nombre, edad, peso, estatura):
        self._nombre = nombre
        self._edad = edad
        self._peso = peso
        self._estatura = estatura


    def get_edad(self):
        return self._edad

    def set_edad(self, nueva_edad):
        if isinstance(nueva_edad, (int, float)) and nueva_edad > 0:
            self._edad = nueva_edad
        else:
            print("Error: La edad debe ser un número positivo.")

    def get_peso(self):
        return self._peso

    def set_peso(self, nuevo_peso):
        if isinstance(nuevo_peso, (int, float)) and nuevo_peso > 0:
            self._peso = nuevo_peso
        else:
            print("Error: El peso debe ser un número positivo.")

    def mostrar_info(self):
        print(f"Nombre: {self._nombre}")
        print(f"Edad: {self._edad}")
        print(f"Peso: {self._peso}")
        print(f"Estatura: {self._estatura}")

def main():
    print("\n=== SISTEMA DE APRENDICES (POO + Encapsulamiento) ===\n")

    aprendiz1 = Aprendiz("Sebastian Soriano", 19, 65, "1.75m")
    aprendiz2 = Aprendiz("Cristina Valderrama", 21, 56, "1.65m")
    aprendiz3 = Aprendiz("Fabio Ramirez", 20, 60, "1.73m")

    print("--- Información inicial ---")
    aprendiz1.mostrar_info()

    print(f"\nConsultando edad: {aprendiz1.get_edad()}")

    print("\nActualizando edad...")
    aprendiz1.set_edad(22)
    print(f"Nueva edad confirmada: {aprendiz1.get_edad()}")

    print(f"\nConsultando peso: {aprendiz1.get_peso()}")

    print("\nActualizando peso...")
    aprendiz1.set_peso(61)

    print(f"Nuevo peso confirmado: {aprendiz1.get_peso()}")
    print("\n--- Información final ---")
    aprendiz1.mostrar_info()

    print("--- Información inicial ---")
    aprendiz2.mostrar_info()

    print(f"\nConsultando edad: {aprendiz2.get_edad()}")

    print("\nActualizando edad...")
    aprendiz2.set_edad(24)
    print(f"Nueva edad confirmada: {aprendiz2.get_edad()}")

    print(f"\nConsultando peso: {aprendiz2.get_peso()}")

    print("\nActualizando peso...")
    aprendiz1.set_peso(58)

    print(f"Nuevo peso confirmado: {aprendiz2.get_peso()}")
    print("\n--- Información final ---")
    aprendiz2.mostrar_info()

    print("--- Información inicial ---")
    aprendiz3.mostrar_info()

    print(f"\nConsultando edad: {aprendiz3.get_edad()}")

    print("\nActualizando edad...")
    aprendiz1.set_edad(23)
    print(f"Nueva edad confirmada: {aprendiz1.get_edad()}")

    print(f"\nConsultando peso: {aprendiz1.get_peso()}")

    print("\nActualizando peso...")
    aprendiz1.set_peso(56)

    print(f"Nuevo peso confirmado: {aprendiz1.get_peso()}")
    print("\n--- Información final ---")
    aprendiz1.mostrar_info()

    print("\nPrograma Finalizado")

if __name__ == "__main__":
    main()
```
