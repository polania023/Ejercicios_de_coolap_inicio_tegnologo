```python
class libro:
    def __init__(self, titulo, precio):
      self._titulo = titulo
      self._precio = precio

    def get_titulo(self):
        return self._titulo

    def set_titulo(self, titulo):
        self._titulo = titulo

    def get_precio(self):
        return self._precio

    def set_precio(self, nuevo_precio):
        if isinstance(nuevo_precio, (int, float)) and nuevo_precio > 0:
            self._precio = nuevo_precio
        else:
            print("ERROR: el precio debe ser un número positivo")

    def mostrar_info(self):
        print(f"titulo: {self._titulo}")
        print(f"precio actual: $ {self._precio:,.2f}")

def main():
  print("\n=== SISTEMA DE LIBROS (POO + encapsulamiento) ===\n")

  libro1 = libro("Python Profesional", 120000)

  print("informacion inicial del libro:")
  libro1.mostrar_info()

  print(f"precio actual:", libro1.get_precio())

  libro1.set_precio(-10)

  libro1.set_precio(120000)

  print("nuevo precio:", libro1.get_precio())

  print("\nPrograma finalizado")

main()
```

```python
class Producto:
    def __init__(self, nombre, talla, color, precio, stock, ubicacion):
        self._nombre = nombre
        self._talla = talla
        self._color = color
        self._precio = precio
        self._stock = stock
        self._ubicacion = ubicacion

    # ===== GETTERS =====
    def get_precio(self):
        return self._precio

    def get_stock(self):
        return self._stock

    def get_ubicacion(self):
        return self._ubicacion

    # ===== SETTERS =====
    def set_precio(self, nuevo_precio):
        if isinstance(nuevo_precio, (int, float)) and nuevo_precio > 0:
            self._precio = nuevo_precio
        else:
            print("Error: El precio debe ser un número positivo.")

    def set_stock(self, nuevo_stock):
        if isinstance(nuevo_stock, int) and nuevo_stock >= 0:
            self._stock = nuevo_stock
        else:
            print("Error: El stock debe ser un número entero positivo.")

    def set_ubicacion(self, nueva_ubicacion):
        if isinstance(nueva_ubicacion, str) and nueva_ubicacion.strip() != "":
            self._ubicacion = nueva_ubicacion
        else:
            print("Error: La ubicación no puede estar vacía.")

    # ===== MOSTRAR INFORMACIÓN =====
    def mostrar_info(self):
        print(f"Nombre: {self._nombre}")
        print(f"Talla: {self._talla}")
        print(f"Color: {self._color}")
        print(f"Precio: ${self._precio}")
        print(f"Stock: {self._stock}")
        print(f"Ubicación: {self._ubicacion}")
        print("-" * 40)


def main():
    print("\n=== SISTEMA DE PRODUCTOS (POO + Encapsulamiento) ===\n")

    # ===== PRODUCTO 1 =====
    producto1 = Producto("Camiseta ", "M", "Negro", 120000, 15, "Bodega A1")

    # ===== PRODUCTO 2 =====
    producto2 = Producto("Pantalón ", "L", "Azul", 180000, 10, "Bodega A2")

    # ===== PRODUCTO 3 =====
    producto3 = Producto("Chaqueta ", "S", "Rojo", 250000, 8, "Bodega B1")

    # ===== PRODUCTO 4 =====
    producto4 = Producto("Sudadera", "XL", "Gris", 200000, 12, "Bodega C3")

    print("--- Información inicial de todos los productos ---\n")
    producto1.mostrar_info()
    producto2.mostrar_info()
    producto3.mostrar_info()
    producto4.mostrar_info()

    print("\nActualizando algunos datos...\n")

    producto1.set_precio(135000)
    producto2.set_stock(15)
    producto3.set_ubicacion("Bodega B2")
    producto4.set_precio(210000)

    print("--- Información final actualizada ---\n")
    producto1.mostrar_info()
    producto2.mostrar_info()
    producto3.mostrar_info()
    producto4.mostrar_info()

    print("\nPrograma Finalizado")


if __name__ == "__main__":
    main()
```

```python
class Libro:
    """
    Uso de decoradores @property para un acceso más limpio
    y controlado a los atributos (encapsulamiento).
    """

    def __init__(self, titulo, precio):
        self.titulo = titulo
        self._precio = precio  # Atributo "protegido"

    @property
    def precio(self):
        """Getter profesional (permite acceder como si fuera un atributo)."""
        return self._precio

    @precio.setter
    def precio(self, nuevo_precio):
        """Setter con validación de seguridad."""
        if isinstance(nuevo_precio, (int, float)) and nuevo_precio > 0:
            self._precio = nuevo_precio
        else:
            print("El precio debe ser un número positivo.")

    def mostrar_info(self):
        print(f"Título: {self.titulo}")
        print(f"Precio: ${self.precio:,.2f}")


def main():

    print("\n=== SISTEMA DE LIBROS (POO - Encapsulamiento) ===\n")

    # Creación del objeto
    libro1 = Libro("Python Profesional", 120000)

    # Mostrar información
    print("Información inicial del libro:")
    libro1.mostrar_info()

    # Acceso directo pero seguro gracias a @property
    print("\nAccediendo al precio con @property:")
    print(libro1.precio)

    # Modificando el precio correctamente
    print("\nModificando el precio...")
    libro1.precio = 150000
    libro1.mostrar_info()

    # Intentando asignar un valor incorrecto
    print("\nIntentando asignar un precio inválido...")
    libro1.precio = -500


if __name__ == "__main__":
    main()
```
